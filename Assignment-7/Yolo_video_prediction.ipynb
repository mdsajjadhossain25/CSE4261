## import YOLO

```python
import cv2
import os
from ultralytics import YOLO
from google.colab.patches import cv2_imshow  # Only needed for Colab visualization
```

## YOLO V11

```python
model = YOLO("yolo11n.pt")
```

## YOLO v11 Prediction on video

```python
# Input and output paths
input_path = '/content/drive/MyDrive/YOLO/input/traffic.mp4'
output_path = '/content/drive/MyDrive/YOLO/output/traffic_yolov11_output.mp4'

# Open the video
cap = cv2.VideoCapture(input_path)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

# Create output directory if it doesn't exist
import os
os.makedirs(os.path.dirname(output_path), exist_ok=True)

# Video Writer
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

# Frame-by-frame inference
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    results = model.predict(source=frame, save=False, conf=0.3, verbose=False)

    # Draw boxes
    annotated_frame = results[0].plot()

    out.write(annotated_frame)

cap.release()
out.release()
print(f"Video saved to: {output_path}")
```
